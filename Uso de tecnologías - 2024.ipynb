```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datai = pd.read_table("EPH_usu_indiv_tic_t424.txt", sep=";")
datai
```

```python
datai.info()
```

```python
usointernet = datai.groupby(datai.IP_III_04).count()
usointernet
```

```python
# Definimos los colores
colores = ['#aec7e8', '#ffbb78', '#e0e0e0']

fig1, ax1 = plt.subplots(figsize=(5, 4)) 

wedges, texts, autotexts = ax1.pie(
    usointernet.CODUSU, 
    labels=None,  # sin etiquetas del gráfico
    colors=colores,
    autopct="%1.0f%%",
    startangle=45, 
    pctdistance=0.75, # Acercamos los porcentajes al centro para evitar bordes
    wedgeprops={"edgecolor":"white", 'linewidth': 2},
    textprops={"fontsize": 12, "color": "#333333"}
)

# Estilo de los porcentajes internos
plt.setp(autotexts, size=10, weight="bold", color="white")

# Título
ax1.set_title("En los últimos meses,¿utilizó internet?", fontsize=12, weight="bold")

# Leyenda lateral
ax1.legend(
    labels=["Si", "No", "Ns/Nr"], 
    title="Referencias",
    loc="center left", 
    bbox_to_anchor=(1, 0, 0.5, 1) #leyenda fuera del gráfico a la derecha
)

plt.tight_layout()
plt.show()
```
