```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datah = pd.read_table("EPH_usu_hog_tic_t424.txt", sep=";")
datah
```

```python
datah.info()
```

```python
# Sobrescribir el dataframe con todas las filas donde IH_II_01 NO es 3
datah = datah[datah["IH_II_01"] != 3]

tienepc = datah.groupby(datah.IH_II_01).count()
tienepc
```

```python
# Definimos los colores: Azul pastel, Naranja pastel, y un Gris neutro para Ns/Nr
colores = ['#aec7e8', '#ffbb78', '#e0e0e0']

fig1, ax1 = plt.subplots(figsize=(4, 4))

wedges, texts, autotexts = ax1.pie(
    tienepc.CODUSU, 
    labels=["Si", "No", "Ns/Nr"],
    colors=colores,
    autopct="%1.0f%%",
    startangle=90,
    wedgeprops={"edgecolor":"white", 'linewidth': 2},
    textprops={"fontsize": 12, "color": "#333333"}
)

# Porcentajes internos
plt.setp(autotexts, size=11, weight="bold", color="white")

# Título
ax1.set_title("¿Tienen computadora en el hogar?", fontsize=12, weight="bold")

plt.tight_layout()
plt.show()
```

```python
internet = datah.groupby(datah.IH_II_02).count()
internet
```

```python
# Sobrescribir el dataframe con todas las filas donde IH_II_02 NO es 4
datah = datah[datah["IH_II_02"] != 4]

internet = datah.groupby(datah.IH_II_02).count()
internet
```

```python
# Definimos los colores
colores = ['#aec7e8', '#ffbb78', '#e0e0e0']

fig1, ax1 = plt.subplots(figsize=(5, 4)) 

wedges, texts, autotexts = ax1.pie(
    internet.CODUSU, 
    labels=None,  # sin etiquetas del gráfico
    colors=colores,
    autopct="%1.0f%%",
    startangle=45, 
    pctdistance=0.75, # Acercamos los porcentajes al centro para evitar bordes
    wedgeprops={"edgecolor":"white", 'linewidth': 2},
    textprops={"fontsize": 12, "color": "#333333"}
)

# Estilo de los porcentajes internos
plt.setp(autotexts, size=10, weight="bold", color="white")

# Título
ax1.set_title("¿Disponen de acceso a Internet?", fontsize=12, weight="bold")

# Leyenda lateral
ax1.legend(
    labels=["Si", "No", "Ns/Nr"], 
    title="Referencias",
    loc="center left", 
    bbox_to_anchor=(1, 0, 0.5, 1) #leyenda fuera del gráfico a la derecha
)

plt.tight_layout()
plt.show()
```
